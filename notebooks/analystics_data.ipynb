```python
import pandas as pd
from pathlib import Path
import numpy as np
```

```python
#Arquivo Archive com o File do Walmart
path_archive = Path.home() /  'Downloads'  / 'archive'
```

```python
for file in path_archive.iterdir():
    if 'Walmart'  in  str(file):
        df_walmart = pd.read_csv(file)
    else:
        print('Arquivo não Encontrado')
```

```python
df_walmart.Unemployment.unique()
```

```python
df_walmart.sort_values('Date',ascending = False)
```

```python
df_walmart['profit_unit'] = df_walmart['CPI'] - df_walmart['Fuel_Price']
```

```python
df_walmart['Status_Temperature'] = np.select(
    [
        df_walmart['Temperature'] >= 50.0,
        (df_walmart['Temperature'] >= 20.0) & (df_walmart['Temperature'] < 50.0),
        df_walmart['Temperature'] < 20.0
    ],
    [
        'High',
        'Medium',
        'Bass',


    ],
    default = 'Unknown'
)
```
